```python
# ---------------- Config ----------------
class Config:
    batch_size = 128
    n_epochs = 100
    lr = 5e-4
    hidden_size = 128
    rnn_layers = 1

    rank = 2
    alpha = 0.2
    noise_std = 0.0

    seq_lengths = (15,15,4,15,10) #rel on, rel off, cue on, cue off, fix off
    seed = 42

class SampleConfig:
    def __init__(self,
                 fc_vals=(1,),
                 b_thetas=(0, 90, 180, 270),      # b_pos angles
                 cue_thetas=(0, 90, 180, 270),      # cue_new angles
                 rel_radius=3,
                 cue_radius=8):
        self.fc_vals = fc_vals
        self.b_thetas = b_thetas
        self.cue_thetas = cue_thetas
        self.rel_radius = rel_radius
        self.cue_radius = cue_radius

COORD_SCALE = 12.0
cfg = Config()
torch.manual_seed(cfg.seed)
np.random.seed(cfg.seed)
random.seed(cfg.seed)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
```

```python
def make_sample(seq_lens, sample_cfg: SampleConfig):
    fc = random.choice(sample_cfg.fc_vals)

    # ---- relation_on positions ----
    g_pos = (0.0, 0.0) # grape always at (0,0)
    b_theta = random.choice(sample_cfg.b_thetas) # b_pos chosen from b_thetas
    b_pos = RF.pol2cart(b_theta, sample_cfg.rel_radius)

    # ---- fc determines target vs cue ----
    if fc == 1:
        target_pos0 = b_pos
        cue_pos0   = g_pos
    elif fc == 2:
        target_pos0 = g_pos
        cue_pos0 = b_pos

    # ---- sample cue_new from cue_thetas ----
    cue_theta = random.choice(sample_cfg.cue_thetas)
    cue_new = RF.pol2cart(cue_theta, sample_cfg.cue_radius)

    # ---- compute final target location ----
    offset = (target_pos0[0] - cue_pos0[0],
              target_pos0[1] - cue_pos0[1])

    target_final = (cue_new[0] + offset[0],
                    cue_new[1] + offset[1])

    # ---------------- Build input sequence ----------------
    seq_len_total = sum(seq_lens)
    feat_dim = 7
    inputs = np.zeros((seq_len_total, feat_dim), dtype=np.float32)

    t = 0
    # 1) relation_on
    for _ in range(seq_lens[0]):
        inputs[t, 0] = 1.0 #b_on
        inputs[t, 1] = 1.0 #g_on
        inputs[t, 2] = 1.0 if fc == 1 else 2.0
        inputs[t, 3] = b_pos[0] / COORD_SCALE
        inputs[t, 4] = b_pos[1] / COORD_SCALE
        inputs[t, 5] = g_pos[0] / COORD_SCALE
        inputs[t, 6] = g_pos[1] / COORD_SCALE
        t += 1

    # 2) relation_off
    for _ in range(seq_lens[1]):
        inputs[t, 2] = 1.0 if fc == 1 else 2.0
        t += 1

    # 3) cue_on (delay 1)
    for _ in range(seq_lens[2]):
        inputs[t, 2] = 1.0 if fc == 1 else 2.0
        if fc == 1: #grape is cue
            inputs[t, 5] = cue_new[0] / COORD_SCALE
            inputs[t, 6] = cue_new[1] / COORD_SCALE
        elif fc == 2: #banana is cue
            inputs[t, 3] = cue_new[0] / COORD_SCALE
            inputs[t, 4] = cue_new[1] / COORD_SCALE
        t += 1

    # 4) cue_off
    for _ in range(seq_lens[3]):
        inputs[t, 2] = 1.0 if fc == 1 else 2.0
        t += 1

    # 5) final cue reappearance
    for _ in range(seq_lens[4]):
        inputs[t, 2] = 0.0
        if fc == 1: #grape is cue
            inputs[t, 5] = cue_new[0] / COORD_SCALE
            inputs[t, 6] = cue_new[1] / COORD_SCALE
        elif fc == 2: #banana is cue
            inputs[t, 3] = cue_new[0] / COORD_SCALE
            inputs[t, 4] = cue_new[1] / COORD_SCALE
        t += 1

    target = np.array([target_final[0] / COORD_SCALE,
                       target_final[1] / COORD_SCALE], dtype=np.float32)

    return inputs, target, fc, b_theta, cue_theta
```

```python
class RelationalDataset(torch.utils.data.Dataset):
    def __init__(self, n_samples, sample_cfg: SampleConfig):
        self.sample_cfg = sample_cfg
        self.samples = [
            make_sample(cfg.seq_lengths, sample_cfg) for _ in range(n_samples)
        ]

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        x, y, fc, b_theta, cue_theta = self.samples[idx]
        meta = torch.tensor([fc, b_theta, cue_theta], dtype=torch.float32)
        return torch.tensor(x, dtype=torch.float32), torch.tensor(y, dtype=torch.float32), meta
```

```python
# ---------------- Low-rank Model ----------------
class RelationalRNN(nn.Module):
    def __init__(
        self,
        input_size=7,
        hidden_size=128,
        num_layers=1,
        rank=2,
        alpha=0.2,
        noise_std=0.0
    ):
        super().__init__()

        if num_layers != 1:
            raise ValueError(
                "Low-rank RNN은 num_layers=1만 사용합니다."
            )

        self.input_size = input_size
        self.hidden_size = hidden_size
        self.num_layers = num_layers
        self.rank = rank
        self.alpha = alpha
        self.noise_std = noise_std

        self.wi = nn.Parameter(
            torch.empty(input_size, hidden_size)
        )

        self.n = nn.Parameter(
            torch.empty(hidden_size, rank)
        )

        self.m = nn.Parameter(
            torch.empty(hidden_size, rank)
        )

        self.bias = nn.Parameter(
            torch.zeros(hidden_size)
        )

        self.readout = nn.Linear(
            hidden_size,
            2
        )

        self.reset_parameters()

    def reset_parameters(self):
        # input weight
        nn.init.xavier_uniform_(self.wi)

        # n, m의 각 열을 직교기저로 초기화
        with torch.no_grad():
            n_random = torch.randn(
                self.hidden_size,
                self.rank
            )

            m_random = torch.randn(
                self.hidden_size,
                self.rank
            )

            n_q, _ = torch.linalg.qr(
                n_random,
                mode="reduced"
            )

            m_q, _ = torch.linalg.qr(
                m_random,
                mode="reduced"
            )

            self.n.copy_(n_q)
            self.m.copy_(m_q)

        nn.init.zeros_(self.bias)

        nn.init.xavier_uniform_(
            self.readout.weight
        )

        nn.init.zeros_(
            self.readout.bias
        )

    def recurrent_step(self, x_t, h_prev):
        input_term = x_t @ self.wi

        # h @ (N M^T)
        latent = h_prev @ self.n
        recurrent_term = latent @ self.m.T

        pre_activation = (
            input_term
            + recurrent_term
            + self.bias
        )

        if self.training and self.noise_std > 0:
            pre_activation = (
                pre_activation
                + self.noise_std
                * torch.randn_like(pre_activation)
            )

        candidate = torch.tanh(
            pre_activation
        )

        h_new = (
            (1.0 - self.alpha) * h_prev
            + self.alpha * candidate
        )

        return h_new

    def run_rnn(self, x):
        batch_size, seq_len, _ = x.shape

        h = x.new_zeros(
            batch_size,
            self.hidden_size
        )

        hidden_states = []

        for t in range(seq_len):
            x_t = x[:, t, :]

            h = self.recurrent_step(
                x_t,
                h
            )

            hidden_states.append(h)

        hidden_seq = torch.stack(
            hidden_states,
            dim=1
        )

        return h, hidden_seq

    def forward(self, x):
        last_hidden, _ = self.run_rnn(x)

        coords = self.readout(
            last_hidden
        )

        return coords

    def forward_with_hidden(self, x):
        last_hidden, hidden_seq = self.run_rnn(x)

        coords = self.readout(
            last_hidden
        )

        return coords, hidden_seq

    def get_recurrent_matrix(self):
        return self.n @ self.m.T
```

```python
class FullRankRelationalRNN(nn.Module):
  def __init__(self,
               input_size=7,
               hidden_size=128,
               num_layers=1,
               alpha=0.2,
               noise_std=0.0):
    super().__init__()

    self.input_size = input_size
    self.hidden_size = hidden_size
    self.num_layers = num_layers
    self.alpha = alpha
    self.noise_std = noise_std

    self.wi = nn.Parameter(torch.empty(input_size, hidden_size))
    self.wrec = nn.Parameter(torch.empty(hidden_size, hidden_size))
    self.bias = nn.Parameter(torch.zeros(hidden_size))
    self.readout = nn.Linear(hidden_size, 2)
    self.reset_parameters()

  def reset_parameters(self):
    nn.init.xavier_uniform_(self.wi)
    nn.init.orthogonal_(self.wrec)
    nn.init.zeros_(self.bias)
    nn.init.xavier_uniform_(self.readout.weight)
    nn.init.zeros_(self.readout.bias)

  def recurrent_step(self, x_t, h_prev):
    input_term = x_t @ self.wi
    recurrent_term = h_prev @ self.wrec

    pre_activation = (input_term + recurrent_term + self.bias)

    if self.training and self.noise_std > 0:
      pre_activation = (pre_activation + self.noise_std * torch.randn_like(pre_activation))

    candidate = torch.tanh(pre_activation)

    h_new = ((1.0 - self.alpha) * h_prev + self.alpha * candidate)

    return h_new

  def run_rnn(self, x):
    batch_size, seq_len, _ = x.shape

    h = x.new_zeros(batch_size, self.hidden_size)

    hidden_states = []

    for t in range(seq_len):
      h = self.recurrent_step(x[:, t, :], h)
      hidden_states.append(h)

    hidden_seq = torch.stack(hidden_states, dim=1)

    return h, hidden_seq

  def forward(self, x):
    last_hidden, _ = self.run_rnn(x)
    return self.readout(last_hidden)

  def forward_with_hidden(self, x):
    last_hidden, hidden_seq = self.run_rnn(x)
    coords = self.readout(last_hidden)
    return coords, hidden_seq

  def get_recurrent_matrix(self):
    return self.wrec
```

```python
# ---------------- Training ----------------
def train_and_evaluate(save_path):
    train_cfg = SampleConfig(
        fc_vals=(1,),
        b_thetas=(0, 90, 180, 270),
        cue_thetas=(0, 90, 180, 270),
    )

    train_ds = RelationalDataset(4000, train_cfg)
    val_ds = RelationalDataset(800, train_cfg)

    train_loader = torch.utils.data.DataLoader(
        train_ds,
        batch_size=cfg.batch_size,
        shuffle=True
    )

    val_loader = torch.utils.data.DataLoader(
        val_ds,
        batch_size=cfg.batch_size,
        shuffle=False
    )

    model = RelationalRNN(
        input_size=7,
        hidden_size=cfg.hidden_size,
        num_layers=cfg.rnn_layers,
        rank=cfg.rank,
        alpha=cfg.alpha,
        noise_std=cfg.noise_std
    ).to(device)

    opt = optim.Adam(model.parameters(), lr=cfg.lr)
    loss_fn = nn.MSELoss()

    def eval_mse(loader):
        model.eval()
        s = 0.0
        n = 0
        with torch.no_grad():
            for xb, yb, info in loader:
                xb = xb.to(device)
                yb = yb.to(device)

                p = model(xb)
                s += ((p - yb) ** 2).sum().item()
                n += yb.numel()

        return s / n

    train_curve = []
    val_curve = []

    for ep in range(1, cfg.n_epochs + 1):

        model.train()
        running = 0.0

        for xb, yb, info in train_loader:

            xb = xb.to(device)
            yb = yb.to(device)

            preds = model(xb)
            loss = loss_fn(preds, yb)

            opt.zero_grad()
            loss.backward()
            opt.step()

            running += loss.item() * xb.size(0)

        train_mse = running / len(train_ds)
        val_mse = eval_mse(val_loader)

        train_curve.append(train_mse)
        val_curve.append(val_mse)

        if ep % 10 == 0 or ep == 1:
            print(
                f"Rank {cfg.rank} | "
                f"Epoch {ep:02d} | "
                f"train {train_mse:.6f} | "
                f"val {val_mse:.6f}"
            )

    torch.save(model.state_dict(), save_path)
    print(f"Model saved to {save_path}")

    return model, val_ds, train_curve, val_curve


# ---------------- Compare ranks ----------------

if __name__ == "__main__":

    ranks = [1, 2, 3, 4]

    results = {}

    for rank in ranks:

        cfg.rank = rank

        save_path = (
            f"models/"
            f"lowrank_rRNN_{cfg.hidden_size}"
            f"_rank{cfg.rank}"
            f"_alpha{cfg.alpha}.pth"
        )

        model, val_ds, train_curve, val_curve = train_and_evaluate(save_path)

        results[rank] = (train_curve, val_curve)


    # ---------- Plot ----------
    plt.figure(figsize=(8,5))

    for rank in ranks:
        _, val_curve = results[rank]

        plt.plot(
            range(1, cfg.n_epochs + 1),
            val_curve,
            label=f"Rank {rank}"
        )

    plt.xlabel("Epoch")
    plt.ylabel("Validation MSE")
    plt.title("Validation Curve")
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
def train_and_evaluate_fullrank(save_path):
    train_cfg = SampleConfig(
        fc_vals=(1,),
        b_thetas=(0, 90, 180, 270),
        cue_thetas=(0, 90, 180, 270),
    )

    train_ds = RelationalDataset(4000, train_cfg)
    val_ds = RelationalDataset(800, train_cfg)

    train_loader = torch.utils.data.DataLoader(
        train_ds,
        batch_size=cfg.batch_size,
        shuffle=True
    )

    val_loader = torch.utils.data.DataLoader(
        val_ds,
        batch_size=cfg.batch_size,
        shuffle=False
    )

    model = FullRankRelationalRNN(
        input_size=7,
        hidden_size=cfg.hidden_size,
        num_layers=cfg.rnn_layers,
        alpha=cfg.alpha,
        noise_std=cfg.noise_std,
    ).to(device)

    opt = optim.Adam(model.parameters(), lr=cfg.lr)
    loss_fn = nn.MSELoss()

    def eval_mse(loader):
        model.eval()
        s = 0.0
        n = 0
        with torch.no_grad():
            for xb, yb, info in loader:
                xb = xb.to(device)
                yb = yb.to(device)

                p = model(xb)
                s += ((p - yb) ** 2).sum().item()
                n += yb.numel()

        return s / n

    train_curve = []
    val_curve = []

    for ep in range(1, cfg.n_epochs + 1):

        model.train()
        running = 0.0

        for xb, yb, info in train_loader:
            xb = xb.to(device)
            yb = yb.to(device)

            preds = model(xb)
            loss = loss_fn(preds, yb)

            opt.zero_grad()
            loss.backward()
            opt.step()

            running += loss.item() * xb.size(0)

        train_mse = running / len(train_ds)
        val_mse = eval_mse(val_loader)

        train_curve.append(train_mse)
        val_curve.append(val_mse)

        if ep % 10 == 0 or ep == 1:
            print(
                f"Epoch {ep:02d}  train {train_mse:.6f}  val {val_mse:.6f}"
            )

    torch.save(model.state_dict(), save_path)

    return model, val_ds, train_curve, val_curve
```

```python
# ============================================================
# 2. Hidden state와 trial metadata 수집
# ============================================================

@torch.no_grad()
def collect_hidden(model, loader, device):
    model.eval()

    xs, ys, preds, hidden, metas = [], [], [], [], []

    for xb, yb, meta in loader:
        xb = xb.to(device)
        yb = yb.to(device)

        pred, hseq = model.forward_with_hidden(xb)

        xs.append(xb.cpu())
        ys.append(yb.cpu())
        preds.append(pred.cpu())
        hidden.append(hseq.cpu())
        metas.append(meta.cpu())

    return {
        "x": torch.cat(xs, dim=0),
        "y": torch.cat(ys, dim=0),
        "pred": torch.cat(preds, dim=0),
        "hidden": torch.cat(hidden, dim=0),
        "meta": torch.cat(metas, dim=0),
    }


rank3_data = collect_hidden(rank3_model, analysis_loader, device)

X = rank3_data["x"]
Y = rank3_data["y"]
PRED = rank3_data["pred"]
H = rank3_data["hidden"]
META = rank3_data["meta"]

test_mse = torch.mean((PRED - Y) ** 2).item()

print("X:", tuple(X.shape))
print("H:", tuple(H.shape))
print("META:", tuple(META.shape))
print(f"analysis MSE: {test_mse:.6f}")
```

```python
# ============================================================
# 3. Epoch 경계와 m-subspace 좌표 계산
# ============================================================

def epoch_boundaries(seq_lengths):
    names = ["relation on", "relation delay", "cue on", "cue delay", "response"]
    ends = np.cumsum(seq_lengths)
    starts = np.concatenate([[0], ends[:-1]])

    return {
        name: (int(start), int(end))
        for name, start, end in zip(names, starts, ends)
    }


EPOCHS = epoch_boundaries(cfg.seq_lengths)
BOUNDARIES = np.cumsum(cfg.seq_lengths)[:-1]

with torch.no_grad():
    Qm, _ = torch.linalg.qr(
        rank3_model.m.detach().cpu(),
        mode="reduced"
    )

Z = H @ Qm

print("Epochs:", EPOCHS)
print("Qm:", tuple(Qm.shape))
print("Z:", tuple(Z.shape))


def add_epoch_lines(ax):
    for t in BOUNDARIES:
        ax.axvline(t - 0.5, linestyle="--", alpha=0.6)
```

```python
# ============================================================
# 4. Rank-3 latent trajectory
# ============================================================

from mpl_toolkits.mplot3d import Axes3D


def plot_rank3_trajectories(
    Z,
    meta,
    group_col=1,
    max_trials_per_condition=12,
    title="Rank-3 trajectories in m-subspace"
):
    if Z.shape[-1] < 3:
        raise ValueError("이 그림은 rank >= 3 모델에서만 그릴 수 있습니다.")

    conditions = sorted(torch.unique(meta[:, group_col]).tolist())
    cmap = plt.get_cmap("tab10")

    fig = plt.figure(figsize=(8, 7))
    ax = fig.add_subplot(111, projection="3d")

    for ci, cond in enumerate(conditions):
        idx = torch.where(meta[:, group_col] == cond)[0]
        idx = idx[:max_trials_per_condition]

        for j, trial_idx in enumerate(idx):
            traj = Z[trial_idx].numpy()

            ax.plot(
                traj[:, 0], traj[:, 1], traj[:, 2],
                color=cmap(ci), alpha=0.35,
                label=f"{int(cond)}°" if j == 0 else None
            )
            ax.scatter(*traj[0, :3], color=cmap(ci), s=18, marker="o")
            ax.scatter(*traj[-1, :3], color=cmap(ci), s=25, marker="x")

    ax.set_xlabel(r"$z_1$")
    ax.set_ylabel(r"$z_2$")
    ax.set_zlabel(r"$z_3$")
    ax.set_title(title)
    ax.legend(title="relation")
    plt.tight_layout()
    plt.show()


plot_rank3_trajectories(
    Z,
    META,
    group_col=1,
    title="Rank-3 trajectories grouped by relation angle"
)
```

```python
# ============================================================
# 5. 조건별 rank-3 latent time course
# ============================================================

def plot_latent_timecourses(Z, meta, group_col, group_name):
    conditions = sorted(torch.unique(meta[:, group_col]).tolist())
    cmap = plt.get_cmap("tab10")
    rank = Z.shape[-1]

    for dim in range(rank):
        plt.figure(figsize=(9, 4.5))

        for ci, cond in enumerate(conditions):
            mask = meta[:, group_col] == cond
            mean = Z[mask, :, dim].mean(dim=0)
            sem = Z[mask, :, dim].std(dim=0) / np.sqrt(max(int(mask.sum()), 1))
            t = np.arange(Z.shape[1])

            plt.plot(
                t,
                mean.numpy(),
                color=cmap(ci),
                linewidth=2,
                label=f"{int(cond)}°"
            )

            plt.fill_between(
                t,
                (mean - sem).numpy(),
                (mean + sem).numpy(),
                color=cmap(ci),
                alpha=0.15
            )

        add_epoch_lines(plt.gca())
        plt.xlabel("Time")
        plt.ylabel(fr"$z_{dim + 1}$")
        plt.title(fr"$z_{dim + 1}$ grouped by {group_name}")
        plt.legend(title=group_name)
        plt.grid(alpha=0.2)
        plt.show()


plot_latent_timecourses(Z, META, group_col=1, group_name="relation")
plot_latent_timecourses(Z, META, group_col=2, group_name="cue")
```

```python
# ============================================================
# 6. m, n, input, readout vector의 cosine overlap
# ============================================================

def cosine_matrix(A, B, eps=1e-8):
    A = A / (torch.linalg.norm(A, dim=1, keepdim=True) + eps)
    B = B / (torch.linalg.norm(B, dim=1, keepdim=True) + eps)
    return A @ B.T


with torch.no_grad():
    m_vecs = rank3_model.m.detach().cpu().T
    n_vecs = rank3_model.n.detach().cpu().T
    input_vecs = rank3_model.wi.detach().cpu()
    output_vecs = rank3_model.readout.weight.detach().cpu()

input_names = [
    "banana_on",
    "grape_on",
    "fc",
    "banana_x",
    "banana_y",
    "grape_x",
    "grape_y",
]

m_names = [f"m{i+1}" for i in range(m_vecs.shape[0])]
n_names = [f"n{i+1}" for i in range(n_vecs.shape[0])]
output_names = ["out_x", "out_y"]

all_vecs = torch.cat([m_vecs, n_vecs, input_vecs, output_vecs], dim=0)
all_names = m_names + n_names + input_names + output_names
overlap = cosine_matrix(all_vecs, all_vecs).numpy()

plt.figure(figsize=(10, 8))
plt.imshow(overlap, vmin=-1, vmax=1, aspect="auto")
plt.colorbar(label="cosine similarity")
plt.xticks(range(len(all_names)), all_names, rotation=90)
plt.yticks(range(len(all_names)), all_names)
plt.title("Connectivity-vector overlap")
plt.tight_layout()
plt.show()

print("n vs. input")
print(np.round(cosine_matrix(n_vecs, input_vecs).numpy(), 3))

print("\nm vs. output")
print(np.round(cosine_matrix(m_vecs, output_vecs).numpy(), 3))
```

```python
# ============================================================
# 8. Hidden activity가 m-subspace에 얼마나 놓이는가
# ============================================================

def subspace_explained_variance(hidden, basis):
    flat = hidden.reshape(-1, hidden.shape[-1])
    centered = flat - flat.mean(dim=0, keepdim=True)
    projection = (centered @ basis) @ basis.T

    total_ss = torch.sum(centered ** 2)
    residual_ss = torch.sum((centered - projection) ** 2)

    return (1.0 - residual_ss / (total_ss + 1e-12)).item()


def participation_ratio(hidden):
    flat = hidden.reshape(-1, hidden.shape[-1])
    flat = flat - flat.mean(dim=0, keepdim=True)

    singular_values = torch.linalg.svdvals(flat)
    variance = singular_values ** 2

    return (
        variance.sum() ** 2
        / (torch.sum(variance ** 2) + 1e-12)
    ).item()


m_explained = subspace_explained_variance(H, Qm)
pr_rank3 = participation_ratio(H)

print(f"Variance explained by span(m): {100*m_explained:.2f}%")
print(f"Hidden-state participation ratio: {pr_rank3:.2f}")

time_explained = []

for t in range(H.shape[1]):
    ht = H[:, t, :]
    ht = ht - ht.mean(dim=0, keepdim=True)
    proj = (ht @ Qm) @ Qm.T

    explained_t = 1.0 - (
        torch.sum((ht - proj) ** 2)
        / (torch.sum(ht ** 2) + 1e-12)
    )

    time_explained.append(explained_t.item())

plt.figure(figsize=(9, 4))
plt.plot(time_explained, linewidth=2)
add_epoch_lines(plt.gca())
plt.ylim(0, 1.05)
plt.xlabel("Time")
plt.ylabel("Variance fraction")
plt.title("Hidden variance explained by recurrent m-subspace")
plt.grid(alpha=0.2)
plt.show()
```

```python
# ============================================================
# 9. Rank-3 vector field의 2차원 단면과 fixed-point 후보
# ============================================================
# 3차원 vector field 전체를 streamplot으로 표시할 수 없으므로,
# 한 latent 축을 대표값에 고정하고 나머지 두 축의 흐름을 그립니다.

@torch.no_grad()
def vector_field_slice_in_m_subspace(
    model,
    Q,
    x_t,
    dims=(0, 1),
    fixed_z=None,
    grid_size=25
):
    model.eval()
    Q = Q.to(device)
    x_t = x_t.to(device).reshape(1, -1)

    d1, d2 = dims
    rank = Q.shape[1]

    if fixed_z is None:
        fixed_z = Z.reshape(-1, rank).median(dim=0).values
    fixed_z = fixed_z.to(device).clone()

    r1 = (float(Z[..., d1].quantile(0.01)), float(Z[..., d1].quantile(0.99)))
    r2 = (float(Z[..., d2].quantile(0.01)), float(Z[..., d2].quantile(0.99)))

    z1 = torch.linspace(*r1, grid_size, device=device)
    z2 = torch.linspace(*r2, grid_size, device=device)
    G1, G2 = torch.meshgrid(z1, z2, indexing="xy")

    grid_z = fixed_z.reshape(1, -1).repeat(G1.numel(), 1)
    grid_z[:, d1] = G1.reshape(-1)
    grid_z[:, d2] = G2.reshape(-1)

    h = grid_z @ Q.T
    x_batch = x_t.expand(h.shape[0], -1)
    h_next = model.recurrent_step(x_batch, h)

    dz_all = (h_next - h) @ Q
    dz = dz_all[:, [d1, d2]]
    speed = torch.linalg.norm(dz_all, dim=1)

    return (
        G1.cpu().numpy(), G2.cpu().numpy(),
        dz[:, 0].reshape(grid_size, grid_size).cpu().numpy(),
        dz[:, 1].reshape(grid_size, grid_size).cpu().numpy(),
        speed.reshape(grid_size, grid_size).cpu().numpy(),
        fixed_z.cpu()
    )


def representative_input(epoch_name, relation_angle=0, cue_angle=0):
    mask = ((META[:, 1] == relation_angle) & (META[:, 2] == cue_angle))
    if not torch.any(mask):
        raise ValueError("해당 조건의 trial이 없습니다.")
    start, end = EPOCHS[epoch_name]
    return X[mask, start:end, :].mean(dim=(0, 1))


def plot_vector_field_slice(
    epoch_name,
    dims=(0, 1),
    relation_angle=0,
    cue_angle=0
):
    x_t = representative_input(epoch_name, relation_angle, cue_angle)

    mask = ((META[:, 1] == relation_angle) & (META[:, 2] == cue_angle))
    mean_traj = Z[mask].mean(dim=0)
    fixed_z = mean_traj.mean(dim=0)

    G1, G2, U, V, SPEED, fixed_z = vector_field_slice_in_m_subspace(
        rank3_model, Qm, x_t, dims=dims, fixed_z=fixed_z, grid_size=27
    )

    d1, d2 = dims
    plt.figure(figsize=(7, 6))
    plt.streamplot(G1, G2, U, V, density=1.1)

    threshold = np.quantile(SPEED, 0.05)
    candidate = SPEED <= threshold
    plt.scatter(G1[candidate], G2[candidate], s=18,
                label="slow / fixed-point candidates")

    mean_traj_np = mean_traj.numpy()
    plt.plot(mean_traj_np[:, d1], mean_traj_np[:, d2], linewidth=3,
             label="observed mean trajectory")
    plt.scatter(mean_traj_np[0, d1], mean_traj_np[0, d2], marker="o", s=50)
    plt.scatter(mean_traj_np[-1, d1], mean_traj_np[-1, d2], marker="x", s=60)

    fixed_dim = ({0, 1, 2} - {d1, d2}).pop()
    plt.xlabel(fr"$z_{d1+1}$")
    plt.ylabel(fr"$z_{d2+1}$")
    plt.title(
        f"Vector-field slice: {epoch_name}\n"
        f"z{fixed_dim+1} fixed at {fixed_z[fixed_dim]:.2f}"
    )
    plt.axis("equal")
    plt.legend()
    plt.grid(alpha=0.2)
    plt.show()


# 세 좌표평면을 모두 확인할 수 있습니다.
for dims in [(0, 1), (0, 2), (1, 2)]:
    plot_vector_field_slice("cue delay", dims=dims, relation_angle=0, cue_angle=0)
```

```python
# ============================================================
# 10. Rank-3 subspace perturbation 분석
# ============================================================

def make_orthogonal_direction(Q, hidden_size, device):
    v = torch.randn(hidden_size, device=device)
    v = v - Q @ (Q.T @ v)
    return v / (torch.linalg.norm(v) + 1e-12)


@torch.no_grad()
def run_with_hidden_perturbation(model, x, perturb_time, direction, magnitude):
    model.eval()
    batch_size, seq_len, _ = x.shape
    h = x.new_zeros(batch_size, model.hidden_size)
    hidden_seq = []

    for t in range(seq_len):
        h = model.recurrent_step(x[:, t, :], h)
        if t == perturb_time:
            h = h + magnitude * direction.reshape(1, -1)
        hidden_seq.append(h)

    hidden_seq = torch.stack(hidden_seq, dim=1)
    return model.readout(h), hidden_seq


Qm_device = Qm.to(device)
axis_directions = [Qm_device[:, i] for i in range(ANALYSIS_RANK)]
orth_direction = make_orthogonal_direction(Qm_device, cfg.hidden_size, device)

perturb_time = EPOCHS["cue delay"][0] + 3
perturb_magnitudes = np.linspace(0.0, 2.0, 9)
x_batch = X[:256].to(device)

with torch.no_grad():
    base_pred, base_h = rank3_model.forward_with_hidden(x_batch)

output_changes = {f"m-axis {i+1}": [] for i in range(ANALYSIS_RANK)}
output_changes["orthogonal"] = []
state_changes = {k: [] for k in output_changes}

directions = {f"m-axis {i+1}": d for i, d in enumerate(axis_directions)}
directions["orthogonal"] = orth_direction

for magnitude in perturb_magnitudes:
    for name, direction in directions.items():
        pred, hseq = run_with_hidden_perturbation(
            rank3_model, x_batch, perturb_time, direction, float(magnitude)
        )
        output_changes[name].append(
            torch.linalg.norm(pred - base_pred, dim=1).mean().item()
        )
        state_changes[name].append(
            torch.linalg.norm(hseq[:, -1] - base_h[:, -1], dim=1).mean().item()
        )

plt.figure(figsize=(8, 5))
for name, values in output_changes.items():
    plt.plot(perturb_magnitudes, values, marker="o", label=name)
plt.xlabel("Perturbation magnitude")
plt.ylabel("Final output change")
plt.title("Rank-3 output sensitivity")
plt.legend()
plt.grid(alpha=0.2)
plt.show()

plt.figure(figsize=(8, 5))
for name, values in state_changes.items():
    plt.plot(perturb_magnitudes, values, marker="o", label=name)
plt.xlabel("Perturbation magnitude")
plt.ylabel("Final hidden-state change")
plt.title("Rank-3 recovery after perturbation")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```
