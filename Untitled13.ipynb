```python
# ---------------- Config ----------------
class Config:
    batch_size = 128
    n_epochs = 200
    lr = 5e-4
    hidden_size = 128
    rnn_layers = 1

    rank = 2
    alpha = 0.2
    noise_std = 0.0

    seq_lengths = (15,15,4,15,10) #rel on, rel off, cue on, cue off, fix off
    seed = 42

class SampleConfig:
    def __init__(self,
                 fc_vals=(1,),
                 b_thetas=(0, 90, 180, 270),      # b_pos angles
                 cue_thetas=(0, 90, 180, 270),      # cue_new angles
                 rel_radius=3,
                 cue_radius=8):
        self.fc_vals = fc_vals
        self.b_thetas = b_thetas
        self.cue_thetas = cue_thetas
        self.rel_radius = rel_radius
        self.cue_radius = cue_radius

COORD_SCALE = 12.0
cfg = Config()
torch.manual_seed(cfg.seed)
np.random.seed(cfg.seed)
random.seed(cfg.seed)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
```

```python
def make_sample(seq_lens, sample_cfg: SampleConfig):
    fc = random.choice(sample_cfg.fc_vals)

    # ---- relation_on positions ----
    g_pos = (0.0, 0.0) # grape always at (0,0)
    b_theta = random.choice(sample_cfg.b_thetas) # b_pos chosen from b_thetas
    b_pos = RF.pol2cart(b_theta, sample_cfg.rel_radius)

    # ---- fc determines target vs cue ----
    if fc == 1:
        target_pos0 = b_pos
        cue_pos0   = g_pos
    elif fc == 2:
        target_pos0 = g_pos
        cue_pos0 = b_pos

    # ---- sample cue_new from cue_thetas ----
    cue_theta = random.choice(sample_cfg.cue_thetas)
    cue_new = RF.pol2cart(cue_theta, sample_cfg.cue_radius)

    # ---- compute final target location ----
    offset = (target_pos0[0] - cue_pos0[0],
              target_pos0[1] - cue_pos0[1])

    target_final = (cue_new[0] + offset[0],
                    cue_new[1] + offset[1])

    # ---------------- Build input sequence ----------------
    seq_len_total = sum(seq_lens)
    feat_dim = 7
    inputs = np.zeros((seq_len_total, feat_dim), dtype=np.float32)

    t = 0
    # 1) relation_on
    for _ in range(seq_lens[0]):
        inputs[t, 0] = 1.0 #b_on
        inputs[t, 1] = 1.0 #g_on
        inputs[t, 2] = 1.0 if fc == 1 else 2.0
        inputs[t, 3] = b_pos[0] / COORD_SCALE
        inputs[t, 4] = b_pos[1] / COORD_SCALE
        inputs[t, 5] = g_pos[0] / COORD_SCALE
        inputs[t, 6] = g_pos[1] / COORD_SCALE
        t += 1

    # 2) relation_off
    for _ in range(seq_lens[1]):
        inputs[t, 2] = 1.0 if fc == 1 else 2.0
        t += 1

    # 3) cue_on (delay 1)
    for _ in range(seq_lens[2]):
        inputs[t, 2] = 1.0 if fc == 1 else 2.0
        if fc == 1: #grape is cue
            inputs[t, 5] = cue_new[0] / COORD_SCALE
            inputs[t, 6] = cue_new[1] / COORD_SCALE
        elif fc == 2: #banana is cue
            inputs[t, 3] = cue_new[0] / COORD_SCALE
            inputs[t, 4] = cue_new[1] / COORD_SCALE
        t += 1

    # 4) cue_off
    for _ in range(seq_lens[3]):
        inputs[t, 2] = 1.0 if fc == 1 else 2.0
        t += 1

    # 5) final cue reappearance
    for _ in range(seq_lens[4]):
        inputs[t, 2] = 0.0
        if fc == 1: #grape is cue
            inputs[t, 5] = cue_new[0] / COORD_SCALE
            inputs[t, 6] = cue_new[1] / COORD_SCALE
        elif fc == 2: #banana is cue
            inputs[t, 3] = cue_new[0] / COORD_SCALE
            inputs[t, 4] = cue_new[1] / COORD_SCALE
        t += 1

    target = np.array([target_final[0] / COORD_SCALE,
                       target_final[1] / COORD_SCALE], dtype=np.float32)

    return inputs, target, fc, b_theta, cue_theta
```

```python
class RelationalDataset(torch.utils.data.Dataset):
    def __init__(self, n_samples, sample_cfg: SampleConfig):
        self.sample_cfg = sample_cfg
        self.samples = [
            make_sample(cfg.seq_lengths, sample_cfg) for _ in range(n_samples)
        ]

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        x, y, fc, b_theta, cue_theta = self.samples[idx]
        meta = torch.tensor([fc, b_theta, cue_theta], dtype=torch.float32)
        return torch.tensor(x, dtype=torch.float32), torch.tensor(y, dtype=torch.float32), meta
```

```python
class RelationalRNN(nn.Module):
  def __init__(
    self,
    input_size=7,
    hidden_size=128,
    num_layers=1,
    rank=2,
    alpha=0.2,
    noise_std=0.0):
    super().__init__()

    if num_layers != 1:
      raise ValueError(
          "이 모델은 단일 층 Low-rank RNN입니다."
          "num_layers는 1이어야 합니다."
      )

    self.input_size = input_size
    self.hidden_size = hidden_size
    self.num_layers = num_layers
    self.rank = rank
    self.alpha = alpha
    self.noise_std = noise_std

    self.wi = nn.Parameter(
        torch.empty(input_size, hidden_size)
    )

    self.m = nn.Parameter(
        torch.empty(hidden_size, rank)
    )

    self.n = nn.Parameter(
        torch.empty(hidden_size, rank)
    )

    self.bias = nn.Parameter(
        torch.zeros(hidden_size)
    )

    self.readout = nn.Linear(hidden_size, 2)

    self.reset_parameter()

  def reset_parameter(self):
    nn.init.xavier_uniform_(self.wi)
    std = 1.0 / np.sqrt(self.hidden_size)

    nn.init.normal_(
        self.m,
        mean=0.0,
        std=std
    )

    nn.init.normal_(
        self.n,
        mean=0.0,
        std=std
    )

    nn.init.zeros_(self.bias)

    nn.init.xavier_uniform_(self.readout.weight)
    nn.init.zeros_(self.readout.bias)

  def recurrent_step(self, x_t, h_prev):
    input_term = x_t @ self.wi
    latent = h_prev @ self.n

    recurrent_term = latent @ self.m.T

    pre_activation = (
        input_term
        + recurrent_term
        + self.bias
    )

    if self.training and self.noise_std > 0:
      noise = torch.randn_like(pre_activation)
      pre_activation = (
          pre_activation
          + self.noise_std * noise
      )

    candidate = torch.tanh(pre_activation)

    h_new = (
        (1.0 - self.alpha) * h_prev
        + self.alpha * candidate
    )

    return h_new

  def run_rnn(self, x):
    batch_size, seq_len, _ = x.shape

    h = x.new_zeros(
        batch_size,
        self.hidden_size
    )

    hidden_states = []

    for t in range(seq_len):
      x_t = x[:, t, :]

      h = self.recurrent_step(
          x_t=x_t,
          h_prev=h
      )

      hidden_states.append(h)

    hidden_seq = torch.stack(
        hidden_states,
        dim=1
    )

    return h, hidden_seq

  def forward(self, x):
    last_hidden, hidden_seq = self.run_rnn(x)

    coords = self.readout(last_hidden)

    return coords

  def forward_with_hidden(self, x):
    last_hidden, hidden_seq = self.run_rnn(x)

    coords = self.readout(last_hidden)

    return coords, hidden_seq

  def get_recurrent_matrix(self):
    return self.m @ self.n.T
```

```python
# ---------------- Training ----------------
def train_and_evaluate(save_path):
    train_cfg = SampleConfig(
        fc_vals=(1,),
        b_thetas=(0, 90, 180, 270),
        cue_thetas=(0, 90, 180, 270),
    )
    train_ds = RelationalDataset(4000, train_cfg)
    val_ds = RelationalDataset(800,train_cfg)
    train_loader = torch.utils.data.DataLoader(train_ds, batch_size=cfg.batch_size, shuffle=True)
    val_loader = torch.utils.data.DataLoader(val_ds, batch_size=cfg.batch_size, shuffle=False)

    model = RelationalRNN(input_size=7, hidden_size=cfg.hidden_size, num_layers=cfg.rnn_layers, rank=cfg.rank, alpha=cfg.alpha, noise_std=cfg.noise_std).to(device)
    opt = optim.Adam(model.parameters(), lr=cfg.lr)
    loss_fn = nn.MSELoss()

    def eval_mse(loader):
        model.eval()
        s = 0.0
        n = 0
        with torch.no_grad():
            for xb, yb, info in loader:
                xb, yb = xb.to(device), yb.to(device)
                p = model(xb)
                s += ((p - yb) ** 2).sum().item()
                n += yb.numel()
        return s / n

    train_curve = []
    val_curve = []
    for ep in range(1, cfg.n_epochs + 1):
        model.train()
        running = 0.0
        for xb, yb, info in train_loader:
            xb, yb = xb.to(device), yb.to(device)
            preds = model(xb)
            loss = loss_fn(preds, yb)
            opt.zero_grad()
            loss.backward()
            opt.step()
            running += loss.item() * xb.size(0)
        train_mse = running / len(train_ds)
        val_mse = eval_mse(val_loader)
        train_curve.append(train_mse)
        val_curve.append(val_mse)
        if ep % 10 == 0 or ep == 1:
            print(f"Epoch {ep:02d}  train_mse {train_mse:.6f}  val_mse {val_mse:.6f}")
    # show some examples
    model.eval()
    for i in range(6):
        xb, yb, info = val_ds[i]
        with torch.no_grad():
            p = model(xb.unsqueeze(0).to(device)).cpu().numpy()[0] * COORD_SCALE
        print(f"Example {i}: Pred ({p[0]:6.2f},{p[1]:6.2f})  Target ({yb.numpy()[0]*COORD_SCALE:6.2f},{yb.numpy()[1]*COORD_SCALE:6.2f})")

    # Save model
    torch.save(model.state_dict(), save_path)
    print(f"Model saved to {save_path}")

    return model, val_ds, train_curve, val_curve

save_path = (
    f"models/lowrank_rRNN_16_"
    f"rank{cfg.rank}_alpha{cfg.alpha}.pth"
)
if __name__ == "__main__":
    if not os.path.exists(save_path):
        model, val_ds, train_curve, val_curve = train_and_evaluate(save_path)

        epochs = range(1, cfg.n_epochs + 1)


        plt.figure(figsize=(8,5))
        plt.plot(epochs, train_curve, label="Train MSE")
        plt.plot(epochs, val_curve, label="Validation MSE")
        plt.xlabel("Epoch")
        plt.ylabel("MSE")
        plt.title("Training and Validation MSE")
        plt.legend()
        plt.grid(True)
        plt.show()
    else:
        print("model already exists")
        #load model
        model = RelationalRNN(input_size=7, hidden_size=cfg.hidden_size, num_layers=cfg.rnn_layers, rank=cfg.rank, alpha=cfg.alpha, noise_std=cfg.noise_std)
        model.load_state_dict(torch.load(save_path,map_location=device))
        model.to(device)
        model.eval()

        train_cfg = SampleConfig(
                fc_vals=(1,),
                b_thetas=(0, 90, 180, 270),
                cue_thetas=(0, 90, 180, 270),
            )
        val_ds = RelationalDataset(800,train_cfg)
        val_loader = torch.utils.data.DataLoader(val_ds, batch_size=cfg.batch_size, shuffle=False)
```

```python
def confusion_matrix_strict(model, loader, sample_cfg):
    # correct if the output location is within 0.55 radius window of answer location
    # 0.55 was chosen bc in 8*8 condition, the minimum distance btw sac targets was ~0.57
    model.eval()

    b_angles = list(sample_cfg.b_thetas)
    num_rel = len(b_angles)

    # confusion matrix: rows = true, columns = predicted
    cm = np.zeros((num_rel, num_rel), dtype=int)

    RADIUS_THRESH = 0.55

    with torch.no_grad():
        for xb, yb, meta in loader:
            xb = xb.to(device)
            preds = model(xb).cpu().numpy() * COORD_SCALE
            metas = meta.numpy()

            for i in range(len(preds)):
                p = preds[i]
                fc  = metas[i][0]
                bth = metas[i][1]
                cueth = metas[i][2]

                # True relation angle
                if fc == 1:
                    theta_true = bth
                else:  # fc==2
                    theta_true = (bth - 180) % 360

                # compute true target position
                cue_new = RF.pol2cart(cueth, sample_cfg.cue_radius)
                true_target = cue_new + RF.pol2cart(theta_true, sample_cfg.rel_radius)

                # distance to TRUE target
                dist_to_true = np.linalg.norm(p - true_target)

                # true relation index
                true_idx = b_angles.index(theta_true)

                # -------- NEW DECISION RULE --------
                if dist_to_true < RADIUS_THRESH:
                    # prediction is counted as correct (lands in correct bin)
                    pred_idx = true_idx
                else:
                    # incorrect: assign to nearest WRONG relation (optional)
                    # compute distance to each candidate
                    candidate_positions = [
                        cue_new + RF.pol2cart(theta, sample_cfg.rel_radius)
                        for theta in b_angles
                    ]
                    dists = [np.linalg.norm(p - c) for c in candidate_positions]
                    pred_idx = int(np.argmin(dists))

                cm[true_idx, pred_idx] += 1

    return cm, b_angles
```

```python
def plot_example_trial(model, loader, sample_cfg, trial_index=0):
    model.eval()

    batch_idx = trial_index // cfg.batch_size
    sample_idx = trial_index % cfg.batch_size
    batch_count = 0
    for xb, yb, meta in loader: # get one batch
        if batch_count == batch_idx:
            inputs = xb.cpu().numpy()
            targets = yb.cpu().numpy()
            metas = meta.cpu().numpy()
            break
        batch_count += 1

    # run through model
    with torch.no_grad():
        preds = model(xb.to(device)).cpu().numpy()

    # pick trial
    print('predsshape',preds.shape)
    p  = preds[sample_idx] * COORD_SCALE
    t  = targets[sample_idx] * COORD_SCALE
    fc = metas[sample_idx][0]
    b_theta  = metas[sample_idx][1]
    cue_theta = metas[sample_idx][2]

    # determine relation angle
    if fc == 1:
        r_theta = b_theta
    else:
        r_theta = (b_theta - 180) % 360

    # relation-on fruit positions
    banana_rel = RF.pol2cart(b_theta, sample_cfg.rel_radius)
    grape_rel  = [0,0]

    # cue-on cue location
    cue_new = RF.pol2cart(cue_theta, sample_cfg.cue_radius)

    # correct target
    correct_target = t

    # predicted target
    predicted = p

    # --- PLOT ---
    fig, ax = plt.subplots(figsize=(5,5))

    # fruits at relation-on
    ax.scatter(grape_rel[0], grape_rel[1], color = '#8a22f2', marker = 's', label="Grape (relation-on)")
    ax.scatter(banana_rel[0], banana_rel[1], color = '#f2d022', label="Banana (relation-on)")

    # cue-on position
    if fc == 1:
        cue_color = '#8a22f2'
        target_color = '#f2d022'
    else:
        cue_color = '#f2d022'
        target_color = '#8a22f2'
    ax.scatter(cue_new[0], cue_new[1], color = cue_color, marker = 's', s = 90, label="Cue location (cue-on)")

    # correct and predicted
    ax.scatter(correct_target[0], correct_target[1], color = target_color, s = 90, label="Correct answer")
    ax.scatter(predicted[0], predicted[1], marker='x', color = 'red', s=100, label="Predicted")


    ax.set_ylim([-15,15])
    ax.set_xlim([-15,15])
    ax.set_aspect('equal')
    ax.legend()
    ax.set_title(f"Trial {trial_index} visualization (fc={fc})")
    ax.grid(True)
    plt.show()
```

```python
def collect_hidden_states(model, dataset):
    model.eval()
    all_hidden = []
    all_fc = []
    all_btheta = []
    all_cuetheta = []
    all_x = []
    all_y = []

    loader = torch.utils.data.DataLoader(dataset, batch_size=1, shuffle=False)

    with torch.no_grad():
        for xb, yb, info in loader:
            xb_on_device = xb.to(device)
            _, hidden_seq = model.forward_with_hidden(xb_on_device)
            # hidden_seq: (1, timesteps, hidden_size)
            all_hidden.append(hidden_seq.squeeze(0).cpu().numpy())
            all_fc.append(int(info[0][0]))
            all_btheta.append(int(info[0][1]))
            all_x.append(xb.cpu().numpy()) # Move to CPU and convert to numpy
            all_y.append(yb.cpu().numpy()) # Move to CPU and convert to numpy
            all_cuetheta.append(int(info[0][2]))

    return (
        np.array(all_hidden),
        np.array(all_fc),
        np.array(all_btheta),
        np.array(all_x),
        np.array(all_y),
        np.array(all_cuetheta)
    )

hidden, fc_arr, btheta_arr, x_arr, y_arr, cuetheta_arr = collect_hidden_states(model, val_ds)
```
