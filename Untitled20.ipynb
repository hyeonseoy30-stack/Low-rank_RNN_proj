```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import random
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import pickle
import colorsys
import math
import os
import RNN_functions as RF
import sys
```

```python
# ---------------- Config ----------------
class Config:
    batch_size = 128
    n_epochs = 200
    lr = 5e-4
    hidden_size = 128
    rnn_layers = 1

    rank = 2
    alpha = 0.2
    noise_std = 0.0

    seq_lengths = (15,15,4,15,10) #rel on, rel off, cue on, cue off, fix off
    seed = 42

class SampleConfig:
    def __init__(self,
                 fc_vals=(1,),
                 b_thetas=(0, 90, 180, 270),      # b_pos angles
                 cue_thetas=(0, 90, 180, 270),      # cue_new angles
                 rel_radius=3,
                 cue_radius=8):
        self.fc_vals = fc_vals
        self.b_thetas = b_thetas
        self.cue_thetas = cue_thetas
        self.rel_radius = rel_radius
        self.cue_radius = cue_radius

COORD_SCALE = 12.0
cfg = Config()
torch.manual_seed(cfg.seed)
np.random.seed(cfg.seed)
random.seed(cfg.seed)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
```

```python
def make_sample(seq_lens, sample_cfg: SampleConfig):
    fc = random.choice(sample_cfg.fc_vals)

    # ---- relation_on positions ----
    g_pos = (0.0, 0.0) # grape always at (0,0)
    b_theta = random.choice(sample_cfg.b_thetas) # b_pos chosen from b_thetas
    b_pos = RF.pol2cart(b_theta, sample_cfg.rel_radius)

    # ---- fc determines target vs cue ----
    if fc == 1:
        target_pos0 = b_pos
        cue_pos0   = g_pos
    elif fc == 2:
        target_pos0 = g_pos
        cue_pos0 = b_pos

    # ---- sample cue_new from cue_thetas ----
    cue_theta = random.choice(sample_cfg.cue_thetas)
    cue_new = RF.pol2cart(cue_theta, sample_cfg.cue_radius)

    # ---- compute final target location ----
    offset = (target_pos0[0] - cue_pos0[0],
              target_pos0[1] - cue_pos0[1])

    target_final = (cue_new[0] + offset[0],
                    cue_new[1] + offset[1])

    # ---------------- Build input sequence ----------------
    seq_len_total = sum(seq_lens)
    feat_dim = 7
    inputs = np.zeros((seq_len_total, feat_dim), dtype=np.float32)

    t = 0
    # 1) relation_on
    for _ in range(seq_lens[0]):
        inputs[t, 0] = 1.0 #b_on
        inputs[t, 1] = 1.0 #g_on
        inputs[t, 2] = 1.0 if fc == 1 else 2.0
        inputs[t, 3] = b_pos[0] / COORD_SCALE
        inputs[t, 4] = b_pos[1] / COORD_SCALE
        inputs[t, 5] = g_pos[0] / COORD_SCALE
        inputs[t, 6] = g_pos[1] / COORD_SCALE
        t += 1

    # 2) relation_off
    for _ in range(seq_lens[1]):
        inputs[t, 2] = 1.0 if fc == 1 else 2.0
        t += 1

    # 3) cue_on (delay 1)
    for _ in range(seq_lens[2]):
        inputs[t, 2] = 1.0 if fc == 1 else 2.0
        if fc == 1: #grape is cue
            inputs[t, 5] = cue_new[0] / COORD_SCALE
            inputs[t, 6] = cue_new[1] / COORD_SCALE
        elif fc == 2: #banana is cue
            inputs[t, 3] = cue_new[0] / COORD_SCALE
            inputs[t, 4] = cue_new[1] / COORD_SCALE
        t += 1

    # 4) cue_off
    for _ in range(seq_lens[3]):
        inputs[t, 2] = 1.0 if fc == 1 else 2.0
        t += 1

    # 5) final cue reappearance
    for _ in range(seq_lens[4]):
        inputs[t, 2] = 0.0
        if fc == 1: #grape is cue
            inputs[t, 5] = cue_new[0] / COORD_SCALE
            inputs[t, 6] = cue_new[1] / COORD_SCALE
        elif fc == 2: #banana is cue
            inputs[t, 3] = cue_new[0] / COORD_SCALE
            inputs[t, 4] = cue_new[1] / COORD_SCALE
        t += 1

    target = np.array([target_final[0] / COORD_SCALE,
                       target_final[1] / COORD_SCALE], dtype=np.float32)

    return inputs, target, fc, b_theta, cue_theta
```

```python
class RelationalDataset(torch.utils.data.Dataset):
    def __init__(self, n_samples, sample_cfg: SampleConfig):
        self.sample_cfg = sample_cfg
        self.samples = [
            make_sample(cfg.seq_lengths, sample_cfg) for _ in range(n_samples)
        ]

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        x, y, fc, b_theta, cue_theta = self.samples[idx]
        meta = torch.tensor([fc, b_theta, cue_theta], dtype=torch.float32)
        return torch.tensor(x, dtype=torch.float32), torch.tensor(y, dtype=torch.float32), meta
```

```python
# ---------------- Low-rank Model ----------------
class RelationalRNN(nn.Module):
    def __init__(
        self,
        input_size=7,
        hidden_size=128,
        num_layers=1,
        rank=2,
        alpha=0.2,
        noise_std=0.0
    ):
        super().__init__()

        if num_layers != 1:
            raise ValueError(
                "Low-rank RNN은 num_layers=1만 사용합니다."
            )

        self.input_size = input_size
        self.hidden_size = hidden_size
        self.num_layers = num_layers
        self.rank = rank
        self.alpha = alpha
        self.noise_std = noise_std

        self.wi = nn.Parameter(
            torch.empty(input_size, hidden_size)
        )

        self.n = nn.Parameter(
            torch.empty(hidden_size, rank)
        )

        self.m = nn.Parameter(
            torch.empty(hidden_size, rank)
        )

        self.bias = nn.Parameter(
            torch.zeros(hidden_size)
        )

        self.readout = nn.Linear(
            hidden_size,
            2
        )

        self.reset_parameters()

    def reset_parameters(self):
        # input weight
        nn.init.xavier_uniform_(self.wi)

        # n, m의 각 열을 직교기저로 초기화
        with torch.no_grad():
            n_random = torch.randn(
                self.hidden_size,
                self.rank
            )

            m_random = torch.randn(
                self.hidden_size,
                self.rank
            )

            n_q, _ = torch.linalg.qr(
                n_random,
                mode="reduced"
            )

            m_q, _ = torch.linalg.qr(
                m_random,
                mode="reduced"
            )

            self.n.copy_(n_q)
            self.m.copy_(m_q)

        nn.init.zeros_(self.bias)

        nn.init.xavier_uniform_(
            self.readout.weight
        )

        nn.init.zeros_(
            self.readout.bias
        )

    def recurrent_step(self, x_t, h_prev):
        input_term = x_t @ self.wi

        # h @ (N M^T)
        latent = h_prev @ self.n
        recurrent_term = latent @ self.m.T

        pre_activation = (
            input_term
            + recurrent_term
            + self.bias
        )

        if self.training and self.noise_std > 0:
            pre_activation = (
                pre_activation
                + self.noise_std
                * torch.randn_like(pre_activation)
            )

        candidate = torch.tanh(
            pre_activation
        )

        h_new = (
            (1.0 - self.alpha) * h_prev
            + self.alpha * candidate
        )

        return h_new

    def run_rnn(self, x):
        batch_size, seq_len, _ = x.shape

        h = x.new_zeros(
            batch_size,
            self.hidden_size
        )

        hidden_states = []

        for t in range(seq_len):
            x_t = x[:, t, :]

            h = self.recurrent_step(
                x_t,
                h
            )

            hidden_states.append(h)

        hidden_seq = torch.stack(
            hidden_states,
            dim=1
        )

        return h, hidden_seq

    def forward(self, x):
        last_hidden, _ = self.run_rnn(x)

        coords = self.readout(
            last_hidden
        )

        return coords

    def forward_with_hidden(self, x):
        last_hidden, hidden_seq = self.run_rnn(x)

        coords = self.readout(
            last_hidden
        )

        return coords, hidden_seq

    def get_recurrent_matrix(self):
        return self.n @ self.m.T
```

```python
class FullRankRelationalRNN(nn.Module):
  def __init__(self,
               input_size=7,
               hidden_size=128,
               num_layers=1,
               alpha=0.2,
               noise_std=0.0):
    super().__init__()

    self.input_size = input_size
    self.hidden_size = hidden_size
    self.num_layers = num_layers
    self.alpha = alpha
    self.noise_std = noise_std

    self.wi = nn.Parameter(torch.empty(input_size, hidden_size))
    self.wrec = nn.Parameter(torch.empty(hidden_size, hidden_size))
    self.bias = nn.Parameter(torch.zeros(hidden_size))
    self.readout = nn.Linear(hidden_size, 2)
    self.reset_parameters()

  def reset_parameters(self):
    nn.init.xavier_uniform_(self.wi)
    nn.init.orthogonal_(self.wrec)
    nn.init.zeros_(self.bias)
    nn.init.xavier_uniform_(self.readout.weight)
    nn.init.zeros_(self.readout.bias)

  def recurrent_step(self, x_t, h_prev):
    input_term = x_t @ self.wi
    recurrent_term = h_prev @ self.wrec

    pre_activation = (input_term + recurrent_term + self.bias)

    if self.training and self.noise_std > 0:
      pre_activation = (pre_activation + self.noise_std * torch.randn_like(pre_activation))

    candidate = torch.tanh(pre_activation)

    h_new = ((1.0 - self.alpha) * h_prev + self.alpha * candidate)

    return h_new

  def run_rnn(self, x):
    batch_size, seq_len, _ = x.shape

    h = x.new_zeros(batch_size, self.hidden_size)

    hidden_states = []

    for t in range(seq_len):
      h = self.recurrent_step(x[:, t, :], h)
      hidden_states.append(h)

    hidden_seq = torch.stack(hidden_states, dim=1)

    return h, hidden_seq

  def forward(self, x):
    last_hidden, _ = self.run_rnn(x)
    return self.readout(last_hidden)

  def forward_with_hidden(self, x):
    last_hidden, hidden_seq = self.run_rnn(x)
    coords = self.readout(last_hidden)
    return coords, hidden_seq

  def get_recurrent_matrix(self):
    return self.wrec
```

```python
# ---------------- Training ----------------
def train_and_evaluate(save_path):
    train_cfg = SampleConfig(
        fc_vals=(1,),
        b_thetas=(0, 90, 180, 270),
        cue_thetas=(0, 90, 180, 270),
    )

    train_ds = RelationalDataset(4000, train_cfg)
    val_ds = RelationalDataset(800, train_cfg)

    train_loader = torch.utils.data.DataLoader(
        train_ds,
        batch_size=cfg.batch_size,
        shuffle=True
    )

    val_loader = torch.utils.data.DataLoader(
        val_ds,
        batch_size=cfg.batch_size,
        shuffle=False
    )

    model = RelationalRNN(
        input_size=7,
        hidden_size=cfg.hidden_size,
        num_layers=cfg.rnn_layers,
        rank=cfg.rank,
        alpha=cfg.alpha,
        noise_std=cfg.noise_std
    ).to(device)

    opt = optim.Adam(model.parameters(), lr=cfg.lr)
    loss_fn = nn.MSELoss()

    def eval_mse(loader):
        model.eval()
        s = 0.0
        n = 0
        with torch.no_grad():
            for xb, yb, info in loader:
                xb = xb.to(device)
                yb = yb.to(device)

                p = model(xb)
                s += ((p - yb) ** 2).sum().item()
                n += yb.numel()

        return s / n

    train_curve = []
    val_curve = []

    for ep in range(1, cfg.n_epochs + 1):

        model.train()
        running = 0.0

        for xb, yb, info in train_loader:

            xb = xb.to(device)
            yb = yb.to(device)

            preds = model(xb)
            loss = loss_fn(preds, yb)

            opt.zero_grad()
            loss.backward()
            opt.step()

            running += loss.item() * xb.size(0)

        train_mse = running / len(train_ds)
        val_mse = eval_mse(val_loader)

        train_curve.append(train_mse)
        val_curve.append(val_mse)

        if ep % 10 == 0 or ep == 1:
            print(
                f"Rank {cfg.rank} | "
                f"Epoch {ep:02d} | "
                f"train {train_mse:.6f} | "
                f"val {val_mse:.6f}"
            )

    torch.save(model.state_dict(), save_path)
    print(f"Model saved to {save_path}")

    return model, val_ds, train_curve, val_curve


# ---------------- Compare ranks ----------------

if __name__ == "__main__":

    ranks = [1, 2, 4, 8]

    results = {}

    for rank in ranks:

        cfg.rank = rank

        save_path = (
            f"models/"
            f"lowrank_rRNN_{cfg.hidden_size}"
            f"_rank{cfg.rank}"
            f"_alpha{cfg.alpha}.pth"
        )

        model, val_ds, train_curve, val_curve = train_and_evaluate(save_path)

        results[rank] = (train_curve, val_curve)


    # ---------- Plot ----------
    plt.figure(figsize=(8,5))

    for rank in ranks:
        _, val_curve = results[rank]

        plt.plot(
            range(1, cfg.n_epochs + 1),
            val_curve,
            label=f"Rank {rank}"
        )

    plt.xlabel("Epoch")
    plt.ylabel("Validation MSE")
    plt.title("Validation Curve")
    plt.legend()
    plt.grid(True)
    plt.show()
```
